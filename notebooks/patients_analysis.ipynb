```python

from pathlib import Path

# Créer le dossier principal du projet
projet = Path("HospitalDataAnalysis")

# Créer les sous-dossiers
(projet / "data").mkdir(parents=True, exist_ok=True)
(projet / "notebooks").mkdir(exist_ok=True)
(projet / "scripts").mkdir(exist_ok=True)

print("Structure du projet créée avec succès !")
```

```python

import csv

fichier = projet / "data" / "patients.csv"

patients = [
    ["PatientID", "Age", "Sex", "BloodPressure", "Cholesterol", "Diagnosis"],
    ["P001", 25, "Male", 120, 180, "Healthy"],
    ["P002", 52, "Female", 145, 240, "At Risk"],
    ["P003", 38, "Male", 130, 210, "Healthy"],
    ["P004", 67, "Female", 160, 260, "At Risk"],
    ["P005", 45, "Male", 125, 195, "Healthy"],
    ["P006", 30, "Female", 118, 175, "Healthy"],
    ["P007", 59, "Male", 150, 245, "At Risk"],
    ["P008", 41, "Female", 135, 220, "At Risk"]
]

with open(fichier, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerows(patients)

print("Fichier patients.csv créé !")
```

```python

import pandas as pd

df = pd.read_csv("HospitalDataAnalysis/data/patients.csv")

df.head()
```

```python

# Statistiques des colonnes numériques
df.describe()
```

```python

# Compter les patients par sexe
df["Sex"].value_counts()
```

```python

# Compter les patients par diagnostic
df["Diagnosis"].value_counts()
```

```python

import matplotlib.pyplot as plt

plt.plot(df["Age"], df["Cholesterol"], marker="o")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.title("Cholesterol vs Age")
plt.show()
```
